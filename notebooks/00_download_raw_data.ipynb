# 00 · Download raw data
Downloads the IBM AML `HI-Small_Trans.csv` from Kaggle and stores it in the `raw` landing Volume. The Kaggle API token is typed in at runtime and never saved in the notebook.

```python
%pip install kagglehub
```

```python
import os, shutil
from getpass import getpass
import kagglehub

os.environ["KAGGLE_API_TOKEN"] = getpass("Kaggle API token: ")

spark.sql("CREATE SCHEMA IF NOT EXISTS workspace.aml_lakehouse")
spark.sql("CREATE VOLUME IF NOT EXISTS workspace.aml_lakehouse.raw")

path = kagglehub.dataset_download(
    "ealtman2019/ibm-transactions-for-anti-money-laundering-aml",
    path="HI-Small_Trans.csv",
)
shutil.copy(path, "/Volumes/workspace/aml_lakehouse/raw/HI-Small_Trans.csv")
display(dbutils.fs.ls("/Volumes/workspace/aml_lakehouse/raw/"))
```
